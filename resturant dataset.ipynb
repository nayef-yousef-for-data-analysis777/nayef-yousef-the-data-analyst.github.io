```python
import pandas as pd
order_df =pd.read_csv(r"C:\Users\CS\Downloads\Compressed\Restaurant_Orders\Restaurant Orders\order_details.csv")
print(order_df)
```

```python
menu_df = pd.read_csv(r"C:\Users\CS\Downloads\Compressed\Restaurant_Orders\Restaurant Orders\menu_items.csv")
print(menu_df)
```

```python
print(order_df.isnull().sum())
print(menu_df.isnull().sum())

```

```python
order_df.dropna(inplace = True)
print(order_df.isnull().sum())
```

```python
print(order_df.dtypes)
print(menu_df.dtypes)
```

```python
order_df['item_id']=order_df['item_id'].convert_dtypes(convert_integer=True)
```

```python
print(order_df.dtypes)
```

```python
menu_df=menu_df.rename(columns={'menu_item_id' : 'item_id'})
print(menu_df)
```

```python
resturant_table = pd.merge(order_df , menu_df , on = 'item_id')
print(resturant_table)
```

```python
print(resturant_table.isnull().sum())
print(resturant_table.dtypes)
```

```python
#What were the least and most ordered items? What categories were they in?
ordered_item = resturant_table.groupby(['category'])['item_name'].value_counts()
print(ordered_item.head(5))
print(ordered_item.tail(5))
```

```python
#What do the highest spend orders look like? Which items did they buy and how much did they spend?
high_spend = resturant_table.groupby(['order_id'])['price'].sum()
sort_high_spend = high_spend.sort_values(ascending=False)
print(sort_high_spend.head(5))
```

```python
#Were there certain times that had more or less orders?
order_time = resturant_table.groupby(['order_time'])['order_id'].count()
print(order_time.head(10))
print(order_time.tail(10))
```

```python
#Which cuisines should we focus on developing more menu items for based on the data?
develop_menu_item = resturant_table.groupby(['item_name'])['order_id'].count()
sort_menu_item = develop_menu_item.sort_values(ascending = True)
print(sort_menu_item)
#we must develop the less ordered items 
```
